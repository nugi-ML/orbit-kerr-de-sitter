# Analisis Data Hasil Simulasi

Notebook ini digunakan HANYA untuk melihat, mem-plot, dan menganalisis data hasil keluaran dari script simulasi (src/simulation.py).
File komputasi berat dipisahkan agar notebook ini cepat dimuat dan digunakan.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# Set parameter spin 'a'
a_val = -0.99

# Set paths
data_dir = '../data/'
results_dir = '../results/'
spin_data_dir = os.path.join(data_dir, f'a_{a_val}')
```

## 1. Membaca Ringkasan (Summary)


```python
summary_file = os.path.join(spin_data_dir, 'deflection_summary.csv')
if os.path.exists(summary_file):
    summary_df = pd.read_csv(summary_file)
    display(summary_df)
else:
    print(f'File {summary_file} belum ada.')
```

## 2. Analisis Fluktuasi Hamiltonian (Konservasi Energi) dari salah satu Lintasan


```python
# Baca salah satu file trajectory, misalnya untuk Lambda = -1e-8
lambda_val = -1e-08
traj_file = os.path.join(spin_data_dir, f'trajectory_lambda_{lambda_val}.csv')

if os.path.exists(traj_file):
    df_traj = pd.read_csv(traj_file)
    
    fig, ax = plt.subplots(figsize=(10,5))
    ax.plot(df_traj['r'], df_traj['H'], color='coral', label=r'Hamiltonian $\mathcal{H}$')
    ax.axhline(0, color='red', linestyle=':', label='Ideal $\mathcal{H}=0$')
    
    # Gunakan symlog karena H fluktuasinya di sekitar nol dengan magnitudo sangat kecil
    ax.set_yscale('symlog', linthresh=1e-12)
    ax.set_title(f'Konservasi Hamiltonian vs r ($\Lambda={lambda_val}$, ={a_val}$)')
    ax.set_xlabel('Koordinat Radial ($)')
    ax.set_ylabel(r'Nilai $\mathcal{H}$')
    ax.grid(True, linestyle='--', alpha=0.6)
    ax.legend()
    plt.show()
else:
    print(f'File {traj_file} belum ada. Jalankan src/simulation.py terlebih dahulu.')
```
